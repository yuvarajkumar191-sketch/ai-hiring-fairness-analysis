---\n<a id='part4'></a>\n## Part 4: AI Impact & Fairness Metrics Analysis\n### **Owner: Member E**\n\n### 4.1 Fairness Metrics: Selection Rates and Disparate Impact\n\nWe compute two key fairness metrics across demographic groups:\n\n1. **Selection Rate:** Proportion of candidates hired per group  \n2. **Disparate Impact Ratio (DIR):** Ratio of selection rates between groups. Values near 1.0 indicate demographic parity; ratios below 0.8 often trigger fairness concerns in audits.

```python
# Helper functions for fairness metrics\n\ndef selection_rate(df, decision_col, group_col):\n    \"\"\"Compute selection rate (hire rate) by group\"\"\"\n    return df.groupby(group_col)[decision_col].mean()\n\ndef disparate_impact(df, decision_col, group_col, ref_group):\n    \"\"\"Compute disparate impact ratio vs. reference group\"\"\"\n    rates = selection_rate(df, decision_col, group_col)\n    return rates / rates[ref_group]\n\nprint('Fairness metric functions defined.')
```

### 5.2 Power and Accountability\n\n- HR and business leaders sometimes believe AI is 'neutral,' but research in HR analytics warns that models trained on unevaluated data can embed organisational and societal bias.  \n- Qualitative interviews with recruiters show they often 'work around' AI tools (e.g., on LinkedIn) when recommendations conflict with their professional judgement, highlighting a tension between efficiency and human agency.  \n- At scale, algorithmic bias can lock thousands of people out of opportunities before they even reach human review.\n\n### 5.3 Governance and Safeguards\n\nBased on our analysis and external research, we propose concrete governance mechanisms:\n\n1. **Regular AI audits and fairness monitoring** of recruitment tools (selection rates, disparate impact across groups).  \n2. **Diverse and representative training data**, including underrepresented groups, to avoid models learning patterns only from majority populations.  \n3. **Human-in-the-loop oversight**, where recruiters remain responsible and can override AI decisions, instead of fully automating hiring.  \n4. **Transparency to candidates** about whether AI is used, how it is debiased, and what rights they have to contest decisions.  \n5. **Accountability frameworks**: clear responsibility for AI decisions, documentation of models, and channels for candidate feedback.
